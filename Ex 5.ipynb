```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
data = {
    "Category": ["A", "B", "C", "D"],
    "Values": [3, 7, 5, 4]
}

df = pd.DataFrame(data)

df.head()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, color='blue', label='Sine Wave')

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot")

plt.legend()
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
values = [3, 7, 5, 4]

plt.bar(categories, values, color='orange')

plt.xlabel("Category")
plt.ylabel("Values")
plt.title("Bar Plot")

plt.show()
```

```python
data = np.random.randn(1000)

plt.hist(
    data,
    bins=20,
    color='purple',
    edgecolor='black'
)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")

plt.show()
```
