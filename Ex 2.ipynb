```python
import pandas as pd
```

```python
data = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha"],
    "Age": [22, 21, 23, 22],
    "Department": ["CSE", "ECE", "IT", "AIDS"],
    "Marks": [85, 90, 88, 92]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

```python
df_csv = pd.read_csv("students.csv")
print(df_csv.head())
```

```python
df.to_excel("students.xlsx", index=False)
print("Excel file created successfully.")
```

```python
df_excel = pd.read_excel("students.xlsx")
print(df_excel.head())
```

```python
import pandas as pd

# Website containing HTML table
url = "https://www.w3schools.com/html/html_tables.asp"

# Read tables from webpage
tables = pd.read_html(url)

# Select the first table
df_web = tables[0]

# Display the imported web data
df_web.head()
```

```python
df_web.to_excel("Web_Data.xlsx", index=False)

print("Data exported successfully to Excel")
```
